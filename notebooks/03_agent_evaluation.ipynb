# Agent Evaluation & Quality Benchmarking

Demonstrates evaluation pipeline, grounding scoring, and hallucination benchmarking.

```python
import asyncio
import sys
sys.path.insert(0, '..')

from app.evaluation.grounding import GroundingEvaluator
from app.evaluation.scoring import QualityScorer

# Evaluation scenarios
scenarios = [
    {
        'name': 'Well-grounded response',
        'query': 'What was Apple revenue in FY2023?',
        'context': ['Apple reported $383.3 billion in total net sales for fiscal year 2023.'],
        'response': 'Apple reported $383.3 billion in revenue for FY2023.',
    },
    {
        'name': 'Poorly grounded response',
        'query': 'What was Apple revenue in FY2023?',
        'context': ['Apple reported $383.3 billion in total net sales for fiscal year 2023.'],
        'response': 'Apple had revenue of $520 billion in FY2024 with record-breaking profits.',
    },
    {
        'name': 'Partial grounding',
        'query': 'Analyze Apple financial performance',
        'context': ['Apple reported $383.3 billion revenue and $97 billion net income in FY2023.'],
        'response': 'Apple had strong performance in FY2023. Revenue was $383.3 billion and profitability remained solid with significant cash generation.',
    },
]

evaluator = GroundingEvaluator(threshold=0.85)
scorer = QualityScorer()

print('Evaluation Results:')
print('=' * 70)

for scenario in scenarios:
    grounding = asyncio.run(evaluator.score_grounding(scenario['query'], scenario['response'], scenario['context']))
    quality = asyncio.run(scorer.score(scenario['query'], scenario['response'], scenario['context']))
    
    print(f"\nScenario: {scenario['name']}")
    print(f"  Grounding Score: {grounding['grounding_score']:.3f} ({'PASS' if grounding['passed'] else 'FAIL'})")
    print(f"  Hallucination Risk: {grounding['hallucination_risk']}")
    print(f"  Overall Quality: {quality['overall_quality']:.3f}")
```

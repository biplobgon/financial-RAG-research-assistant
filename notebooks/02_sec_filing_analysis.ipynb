# SEC Filing Analysis — Financial RAG Demo

This notebook demonstrates SEC filing analysis using the SEC Filing Agent.

```python
import asyncio
import json
import sys
sys.path.insert(0, '..')

# Load synthetic SEC filing data
with open('../data/synthetic/sec_filings_sample.json') as f:
    filings = json.load(f)

print(f'Loaded {len(filings)} synthetic SEC filings')
for filing in filings:
    print(f"  {filing['ticker']} {filing['filing_type']} {filing['fiscal_year']} — {filing['title']}")
```

```python
from app.utils.financial_utils import format_currency, parse_market_cap_tier

# Analyze key financials from synthetic data
print('Key Financial Comparison (FY2023):')
print('-' * 60)
for filing in filings:
    metrics = filing.get('key_financials', {})
    revenue = metrics.get('revenue', 0)
    net_income = metrics.get('net_income', 0)
    margin = (net_income / revenue * 100) if revenue else 0
    print(f"{filing['ticker']:6} | Revenue: {format_currency(revenue):12} | Net Income: {format_currency(net_income):12} | Margin: {margin:.1f}%")
```

```python
# Text analysis on SEC filing content
from app.utils.text_processing import extract_financial_figures, clean_financial_text

for filing in filings[:2]:
    content = filing['content']
    cleaned = clean_financial_text(content)
    figures = extract_financial_figures(content)
    
    print(f"\n{filing['ticker']} {filing['filing_type']}:")
    print(f"  Content length: {len(content)} chars")
    print(f"  Financial figures found: {len(figures)}")
    for fig in figures[:5]:
        from app.utils.financial_utils import format_currency
        print(f"    {fig['raw']} -> {format_currency(fig['value'])}")
```
